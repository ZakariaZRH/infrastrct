```python
%run filter_sysinfo.ipynb
import openpyxl

data = filter_info(collect_system_info())           

clean_data = {}
for k, v in data.items():
    if v is None:
        clean_data[k] = ""
    elif isinstance(v, list):
        clean_data[k] = " | ".join(str(item) for item in v)
    else:
        clean_data[k] = v

wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Systeeminformatie"

columns = list(clean_data.keys())
ws.append(columns)

row_values = [clean_data[k] for k in columns]
ws.append(row_values)

path = f"sysinfo_{clean_data['computernaam']}.xlsx"
wb.save(path)

print("Spreadsheet opgeslagen:", path)
```
